### First example: a quadratic function

Consider the following optimization problem
$$\min_{x \in \mathbb{R}^2} f(x)$$
where
$$f(\mathbf{x}) = \frac{1}{2} \mathbf{x}^\top \mathbf{Q} \mathbf{x} - \mathbf{p}^\top \mathbf{x} \text{ with } \begin{cases} \mathbf{Q} = \begin{bmatrix} 1 & 0 \\ 0 & \eta\end{bmatrix}, \: \eta > 0\\
\mathbf{p} = \begin{bmatrix} 1\\1\end{bmatrix}\end{cases}$$

**Preliminary questions**
1. Recall the expression of $\nabla f(\mathbf{x})$.
2. What is a minimizer of this optimization problem? Is it unique?
3. Recall the principle of gradient descent using pseudo-code.

1. $$∇f(x) = Qx - \mathbf{p}$$

2. A minimizer of this function is $$ x = Q^{-1}p $$
f est strictement convexe donc si elle admet un minimum local, c'est aussi un minimum global et celui ci est unique.

En effet,

$$∀ x,y \in \mathbb{R}$$
$$(∇f(x) - ∇f(y))^\top  (x-y)>0 $$
donc la fonction f est strictement convexe et x est l'unique min global de f.

3. La méthode de descente de gradient consiste à construire une suite $x^{k}$ qui converge vers un min local de la fonction qu'on souhaite etudier.

$∀ k $  $x^{k+1} = x^{k} - \alpha_{k}∇f(x^{k})$

On doit alors choisir une valeur de pas de sorte à controller la convergence et sa vitesse.


**Programming questions**

```python
# question 1
p=np.array([1,1]).T
def f(X,eta=5):
  Q=np.array([[1,0],
   [0,eta]])
  return 1/2*(X.T@Q@X) - p.T@X
```

```python
def F(x,y, eta=5):
  return (1/2*x**2+eta*y**2)-x-y
```

```python
# question 2
x=np.linspace(-1,1,100)
y=np.linspace(-1,1,100)

X,Y = np.meshgrid(x,y)
Z=F(X,Y, eta=5)

plt.pcolormesh(X,Y,Z) # represente seulement des shades de couleurs selon la valeur de la fonction

plt.contour(X,Y,Z) # represente des lignes

plt.contourf(X,Y,Z) # mix des deux

```

```python
# question 3
def grad(X, eta=5):
  p=np.array([1,1]).T
  Q=np.array([[1,0],
   [0,eta]])
  return Q@X-p

def desc_grad_1(x0,alpha, NITR):
  eps = 10**-5
  n=0
  x=x0
  while n<NITR:
    y=x
    x=y-alpha*grad(x,eta=5)
    dist=np.linalg.norm(x-y)/np.linalg.norm(x)
    if dist<eps:
      return x,n
    n=n+1
  return x

# on observe avec alpha = 0.02, une convergence en 28 iterations

def desc_grad_2(x0,alpha, NITR):
  eps = 10**-5
  n=0
  x=x0
  while n<NITR:
    y=x
    x=y-alpha*grad(x,eta=5)
    dist=np.linalg.norm(f(x, eta=5)-f(y, eta=5))/np.linalg.norm(f(x, eta=5))
    if dist<eps:
      return x,n
    n=n+1
  return x

def desc_grad_3(x0,alpha, NITR):
  eps = 10**-5
  n=0
  x=x0
  while n<NITR:
    y=x
    x=y-alpha*grad(x,eta=5)
    dist=np.linalg.norm(grad(x, eta=5))
    if dist<eps:
      return x,n
    n=n+1
  return x

print(desc_grad_1(np.array([0.4,0.1]).T,0.08, 10000))
print(desc_grad_2(np.array([0.4,0.1]).T,0.08, 10000))
print(desc_grad_3(np.array([0.4,0.1]).T,0.08, 10000))

# on remarque qu'avecc un pas alpha plus grand on obtient une vitesse de convergence plus elevée. Ici, pour alpha=0,08 les resultats sont assez rapides
```

```python
# question 4
def desc_grad_optimal(x0, NITR, eta=5):
  eps = 10**-5
  n=0
  x=x0
  Q=np.array([[1,0],
   [0,eta]])
  while n<NITR:
    y=x
    alpha = (grad(x, eta=5).T@grad(x, eta=5))/(grad(x, eta=5).T@Q@grad(x, eta=5))
    x=y-alpha*grad(x,eta=5)
    dist=np.linalg.norm(x-y)/np.linalg.norm(x)
    if dist<eps:
      return x,n
    n=n+1
  return x
desc_grad_optimal(np.array([0.4,0.1]).T, 10000)
```
